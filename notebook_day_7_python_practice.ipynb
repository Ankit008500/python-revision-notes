A function is a reusable block of code that performs a specific task.

Why functions are useful:
- reduce repeated code
- make programs cleaner
- improve readability
- organize logic properly

```python
# Basic function example

def hello():
    print("Hello Mitra")

hello()
```

```python
# Function with parameter

def greet(name):
    print("Hello", name)

greet("Ankit")
greet("Rahul")
```

```python
# Function with return value

def add10(x):
    return x + 10

print(add10(20))
```

```python
# Function with two parameters

def add(a, b):
    return a + b

print(add(45, 65))
```

```python
# Even or odd checker

def even_odd(x):
    if x % 2 == 0:
        print(x, "is even")
    else:
        print(x, "is odd")

evenodd(2883)
evenodd(200)
```

```python
# Function with default argument

def welcome(name="Guest"):
    print("Welcome", name)

welcome()
welcome("Ankit")
```

```python
# Factorial function

def factorial(n):
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result

print(factorial(5))
```

```python
# Mini Project - Student Result Checker

name = input("Enter your name: ")
marks = int(input("Enter your marks: "))

def grade(marks):

    if marks >= 90:
        return "A"

    elif marks >= 75:
        return "B"

    elif marks >= 50:
        return "C"

    else:
        return "Fail"

result = grade(marks)

print("Hello", name)
print("Your grade is:", result)
```

## Common Mistakes and Fixes

### Mistake 1
Using print(function()) when the function already prints output.

Fix:
Call the function directly.

### Mistake 2
Forgetting return when a value is needed later.

Fix:
Use return to send values back.

### Mistake 3
Writing repeated code.

Fix:
Put repeated logic inside functions.

### Mistake 4
Using confusing variable names.

Fix:
Use readable names like:
- marks
- result
- items
- numbers
