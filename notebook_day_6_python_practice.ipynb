# Day 6 - Collections

Today I am learning:
- list
- tuple
- set
- dictionary

Key idea:
- List is ordered and changeable
- Tuple is ordered and not changeable
- Set is unordered and stores unique values only
- Dictionary stores key-value pairs

```python
# List is ordered and mutable
items = [1, "sparta", 3.14, True, 3+5j]

print(items)
print(type(items))

# Indexing
print(items[1])

# Slicing
print(items[1:4])

# Modify item
items[0] = 100
print(items)

# Append item
items.append("mango")
print(items)

# Insert item
items.insert(1, "python")
print(items)

# Remove last item
items.pop()
print(items)

# Sort numbers
numbers = [4, 1, 9, 2, 6]
numbers.sort()
print(numbers)
```

```python
# Tuple is ordered and immutable
languages = ("Python", "Java", "C++")

print(languages)
print(type(languages))
print(languages[0])
print(languages[1:])
```

```python
# Set is unordered and stores unique values only
items = {1, 'a', 1, 2, 3}

print(items)
print(type(items))

# Add new element
items.add("hello")
print(items)

# Remove element
items.remove("a")
print(items)

# Union
s1 = {1, 2, 3}
s2 = {4, 5, 6}
print(s1.union(s2))

# Intersection
s3 = {1, 2, 3, 4}
s4 = {3, 4, 5, 6}
print(s3.intersection(s4))
```

```python
# Dictionary is unordered and stores key-value pairs
fruit = {
    "Apple": 10,
    "Banana": 20
}

print(fruit)
print(type(fruit))

# Access value
print(fruit["Apple"])

# Keys, values, items
print(fruit.keys())
print(fruit.values())
print(fruit.items())

# Add new element
fruit["Mango"] = 30
print(fruit)

# Update value
fruit["Apple"] = 100
print(fruit)

# Update using another dictionary
new_items = {"Orange": 40}
fruit.update(new_items)
print(fruit)

# Remove element
fruit.pop("Banana")
print(fruit)
```

## Now I will practice collections with small examples.

```python
# List practice
fruits = ["apple", "banana", "mango", "grapes"]

print(fruits[0])
print(fruits[1])
print(fruits[0:2])

fruits.append("orange")
fruits.insert(1, "kiwi")
fruits.remove("banana")
print(fruits)
```

```python
# Tuple practice
colors = ("red", "green", "blue", "yellow")

print(colors)
print(colors[2])
print(colors[1:])
```

```python
# Set practice
nums = {1, 2, 2, 3, 4, 4, 5}

print(nums)

nums.add(10)
print(nums)

nums.update([20, 30])
print(nums)

nums.remove(2)
print(nums)
```

```python
# Dictionary practice
# Dictionary is unordered and stores key-value pairs
items = {
    "Apple": 10,
    "Banana": 20
}

print(items)
print(type(items))

# Access value
print(items["Apple"])

# Keys, values, items
print(items.keys())
print(items.values())
print(items.items())

# Add new element
items["Mango"] = 30
print(items)

# Update value
items["Apple"] = 100
print(items)

# Update using another dictionary
new_items = {"Orange": 40}
items.update(new_items)
print(items)

# Remove element
items.pop("Banana")
print(items)
```

## Common Mistakes and Fixes

### Mistake 1
Using list when data should not change.

Fix:
Use tuple when values must stay fixed.

### Mistake 2
Trying to add duplicate values in a set and expecting them to stay.

Fix:
Set keeps only unique values.

### Mistake 3
Using dictionary like a list.

Fix:
Dictionary uses keys, not index positions.

### Mistake 4
Trying to change tuple values.

Fix:
Tuple cannot be changed after creation.

## Summary

- List is ordered and changeable
- Tuple is ordered and not changeable
- Set stores unique values only
- Dictionary stores key-value pairs
- Collections help organize data better
