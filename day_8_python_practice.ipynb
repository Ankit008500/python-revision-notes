## File Handling
File handling is used to read, write and update files.

Important modes:
- r → read
- w → write
- a → append

Important functions:
- open()
- read()
- write()
- close()

## Exception Handling
Exception handling is used to prevent program crashes when errors happen.

Important keywords:
- try
- except
- finally

## Modules
Modules help reuse code written by others.

Examples:
- math
- random

## Import
import is used to use modules in Python.

```python
# Read file example

file = open("sample.txt", "r")

content = file.read()

print(content)

file.close()
```

```python
# Write file example

file = open("sample.txt", "w")

file.write("Hello Python")

file.close()

print("Data written successfully")
```

```python
# Append file example

file = open("sample.txt", "a")

file.write("\nLearning Python Fundamentals")

file.close()

print("Data appended successfully")
```

```python
# Exception handling example

try:

    num = int(input("Enter a number: "))

    print(num)

except:

    print("Invalid input")

finally:

    print("Program finished")
```

```python
# Zero division handling

try:

    a = 10
    b = 0

    print(a / b)

except ZeroDivisionError:

    print("Cannot divide by zero")
```

```python
# Import math module

import math

print(math.sqrt(25))
print(math.factorial(5))
```

```python
# Import random module

import random

print(random.randint(1, 10))
```

```python
# Mini Project - Simple File Notes App

try:

    file = open("notes.txt", "a")

    note = input("Enter your note: ")

    file.write(note + "\n")

    file.close()

    print("Note saved successfully")

except:

    print("Error while saving note")
```
