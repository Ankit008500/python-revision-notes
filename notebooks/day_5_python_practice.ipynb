## Day 5 : Patters 

Today I am learning pattern printing using nested loops.

Key Idea:
- Outer loop → controls rows
- Inner loop → controls columns

If inner loop does not depend on outer loop,
pattern will repeat same line.

```python
# Print star triangle

rows = 5

for i in range(1, rows + 1):
    for j in range(i):
        print("*", end="")
    print()
```

```python
# Print increasing numbers

rows = 5

for i in range(1, rows + 1):
    for j in range(1, i + 1):
        print(j, end="")
    print()
```

```python
# Print decreasing numbers

rows = 5

for i in range(rows, 0, -1):
    for j in range(i, 0, -1):
        print(j, end="")
    print()
```

```python
# Right aligned pyramid pattern
# First loop → spaces
# Second loop → stars

rows = 7

for i in range(1, rows + 1):
    for j in range(rows - i):
        print(" ", end="")
    for k in range(i):
        print("*", end=" ")
    print()
```

## Common Mistakes

1. Inner loop not depending on outer loop
→ Causes same output every row

2. Wrong range logic
→ Pattern shape becomes incorrect

3. Forgetting end=""
→ Output goes to next line

4. Confusion between row and column logic

## Fixes

- Always ask: Does inner loop depend on i?
- Understand pattern shape before coding
- Use print(..., end="") for same line

## Summary

- Patterns are built using nested loops
- Outer loop controls rows
- Inner loop controls columns
- Pattern shape depends on inner loop logic
- Understanding is more important than memorizing
