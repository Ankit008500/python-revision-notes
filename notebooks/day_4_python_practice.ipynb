## Loop

A loop is used to repeat the same code multiple times.

There are two main types:
- for loop
- while loop

```python
# Print numbers from 0 to 4

for i in range(5):
    print(i)
```

## range()

range(5) → 0 to 4  
range(1, 5) → 1 to 4  

The last number is not included.

```python
# Print numbers from 1 to 5

for i in range(1, 6):
    print(i)
```

## while loop

A while loop runs until the condition becomes false.

```python
# Print numbers from 0 to 4 using while loop

count = 0

while count < 5:
    print(count)
    count += 1
```

```python
# Print numbers from 1 to 10

for i in range(1, 11):
    print(i)
```

```python
# Print even numbers from 1 to 10

for i in range(1, 11):
    if i % 2 == 0:
        print(i)
```

```python
# Find sum of numbers from 1 to n

n = int(input("Enter a number: "))  # input used only once
total = 0  # store sum

for i in range(1, n + 1):
    total += i

print("Sum:", total)
```

```python
# Print multiplication table of a number

n = 5  

for i in range(1, 11):
    print(n, "x", i, "=", n * i)
```

```python
# Countdown from 5 to 1

i = 5

while i > 0:
    print(i)
    i -= 1
```
