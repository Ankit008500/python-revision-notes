## What is OOP?
Object Oriented Programming (OOP) is a way of writing code using classes and objects.

## Important Terms

### Class
A class is a blueprint for creating objects.

### Object
An object is created from a class.

### Attributes
Attributes are the properties of an object.

### Methods
Methods are functions inside a class.

### Constructor
A constructor is a special method that runs automatically when an object is created.

### self
self refers to the current object.

```python
# Class and object example

class Phone:
    name = "Samsung"

phone1 = Phone()

print(phone1.name)
```

```python
# Class with attributes and methods

class Phone:
    name = "iPhone"
    color = "Black"

    def make_call(self):
        print("Making a call")

    def play_game(self):
        print("Playing a game")

p1 = Phone()

print(p1.name)
print(p1.color)
p1.make_call()
p1.play_game()
```

```python
# Class with constructor

class Employee:
    def __init__(self, name, age, salary, gender):
        self.name = name
        self.age = age
        self.salary = salary
        self.gender = gender

    def show_details(self):
        print("The name of employee is", self.name)
        print("The age of employee is", self.age)
        print("The salary of employee is", self.salary)
        print("The gender of employee is", self.gender)

e1 = Employee("Raghu", 25, 70000, "Male")
e1.show_details()
```

```python
# Mini project - Student class

class Student:
    def __init__(self, name, age, marks):
        self.name = name
        self.age = age
        self.marks = marks

    def show_result(self):
        if self.marks >= 50:
            print(self.name, "has passed")
        else:
            print(self.name, "has failed")

s1 = Student("Ankit", 20, 78)
s2 = Student("Rahul", 19, 40)

s1.show_result()
s2.show_result()
```

```python
# Mini project - Car class

class Car:
    def __init__(self, brand, mileage, cost):
        self.brand = brand
        self.mileage = mileage
        self.cost = cost

    def show_details(self):
        print("I am a car")
        print("Brand:", self.brand)
        print("Mileage:", self.mileage)
        print("Cost:", self.cost)

c1 = Car("BMW", 100, 2000)
c1.show_details()
```

## Now I will practice classes, objects, methods, and constructors.

Main idea:
- class stores structure
- object stores values
- methods perform actions
- constructor sets initial values

```python
# Practice example

class Mobile:
    def __init__(self, brand, price):
        self.brand = brand
        self.price = price

    def show_brand(self):
        print(self.brand)

    def show_price(self):
        print(self.price)

# creating object 
m1 = Mobile("Redmi", 800)

m1.show_brand()
m1.show_price()
```
